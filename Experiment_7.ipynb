```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

# -----------------------------
# 2. Create the Dataset
# -----------------------------

data = {
    "Student_ID": range(1, 16),

    "Name": [
        "Arun", "Bala", "Chitra", "Divya", "Eshan",
        "Farah", "Gokul", "Hari", "Isha", "Jaya",
        "Kiran", "Latha", "Manoj", "Nisha", "Omkar"
    ],

    "Age": [
        20, 21, 20, 22, 21,
        20, 23, 21, 22, 20,
        21, 22, 20, 21, 22
    ],

    "Gender": [
        "Male", "Male", "Female", "Female", "Male",
        "Female", "Male", "Male", "Female", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],

    "Department": [
        "CSE", "ECE", "CSE", "EEE", "CSE",
        "ECE", "EEE", "CSE", "ECE", "EEE",
        "CSE", "ECE", "CSE", "EEE", "ECE"
    ],

    "Study_Hours": [
        5, 3, 6, 2, 4,
        6, 1, 4, 7, 3,
        5, 6, 4, 5, 3
    ],

    "Attendance": [
        95, 80, 98, 75, 85,
        92, 60, 82, 96, 78,
        90, 94, 88, 91, 79
    ],

    "Marks": [
        85, 72, 90, 65, 78,
        88, 55, 76, 92, 70,
        84, 89, 80, 86, 73
    ],

    "Feedback": [
        "Excellent", "Good", "Excellent", "Average", "Good",
        "Excellent", "Poor", "Good", "Excellent", "Average",
        "Good", "Excellent", "Good", "Excellent", "Average"
    ]
}

df = pd.DataFrame(data)

# Display complete dataset
print("=" * 60)
print("DATASET")
print("=" * 60)
print(df)


# ============================================================
# 3. BASIC DATASET INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("BASIC INFORMATION")
print("=" * 60)

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Records:")
print(df.head())

print("\nLast 5 Records:")
print(df.tail())


# ============================================================
# 4. DATA QUALITY CHECK
# ============================================================

print("\n" + "=" * 60)
print("DATA QUALITY CHECK")
print("=" * 60)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nUnique Values:")
print(df.nunique())


# ============================================================
# 5. QUANTITATIVE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("QUANTITATIVE ANALYSIS")
print("=" * 60)

# Select numerical columns
numerical_columns = [
    "Age",
    "Study_Hours",
    "Attendance",
    "Marks"
]

# Descriptive statistics
print("\nDescriptive Statistics:")
print(df[numerical_columns].describe())

# Mean
print("\nMean:")
print(df[numerical_columns].mean())

# Median
print("\nMedian:")
print(df[numerical_columns].median())

# Standard deviation
print("\nStandard Deviation:")
print(df[numerical_columns].std())

# Minimum and maximum
print("\nMinimum Values:")
print(df[numerical_columns].min())

print("\nMaximum Values:")
print(df[numerical_columns].max())


# ============================================================
# 6. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("CORRELATION MATRIX")
print("=" * 60)

correlation = df[numerical_columns].corr()

print(correlation)


# ============================================================
# 7. QUALITATIVE / CATEGORICAL ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("QUALITATIVE ANALYSIS")
print("=" * 60)

# Gender distribution
print("\nGender Distribution:")
print(df["Gender"].value_counts())

# Department distribution
print("\nDepartment Distribution:")
print(df["Department"].value_counts())

# Feedback distribution
print("\nFeedback Distribution:")
print(df["Feedback"].value_counts())

# Cross-tabulation
print("\nDepartment vs Gender:")
print(pd.crosstab(df["Department"], df["Gender"]))

print("\nDepartment vs Feedback:")
print(pd.crosstab(df["Department"], df["Feedback"]))


# ============================================================
# 8. VISUALIZATION 1 - HISTOGRAM
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    df["Marks"],
    bins=6,
    kde=True,
    color="royalblue"
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()


# ============================================================
# 9. VISUALIZATION 2 - BOX PLOT
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    y=df["Marks"],
    color="lightgreen"
)

plt.title("Box Plot of Student Marks")
plt.ylabel("Marks")
plt.tight_layout()
plt.show()


# ============================================================
# 10. VISUALIZATION 3 - BAR PLOT
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    x="Department",
    data=df,
    hue="Department",
    palette="Set2",
    legend=False
)

plt.title("Number of Students by Department")
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```
